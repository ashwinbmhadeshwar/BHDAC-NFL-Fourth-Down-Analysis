```python
import numpy as np
import pandas as pd
import nflreadpy as nfl
import pyarrow
##importing matplotlib gives an error
```

```python
data = nfl.load_pbp([2022, 2023, 2024, 2025])
df = data.to_pandas()
df.head()
```

```python
fourth = df[(df["down"] == 4) & (df["season_type"] == "REG")]

cols = ["season", "game_id", "posteam", "qtr", "game_seconds_remaining",
        "ydstogo", "yardline_100", "score_differential", "play_type",
        "fourth_down_converted", "fourth_down_failed", "wp", "ep"]
fourth = fourth[cols]
print(fourth.shape)
fourth.head()
```

```python
fourth = fourth.dropna()
fourth.isna().sum()
```

```python
po4 = fourth[(fourth["play_type"] != "punt") & (fourth["play_type"] != "field_goal") & (fourth["play_type"] != "no_play") & (fourth["play_type"] != "qb_kneel")]
po4.head()
```
